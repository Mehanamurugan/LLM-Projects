```python

import json
import speech_recognition as sr
!pip install PyAudio

from gtts import gTTS
from langchain_groq import ChatGroq
from IPython.display import Audio, display
# ==========================================
# LOAD API
# ==========================================

api_key = input("Enter Groq API Key: ")
model_name = input("Enter Model Name: ")

llm = ChatGroq(
    api_key=api_key,
    model=model_name
)

# ==========================================
# CHAT HISTORY
# ==========================================

chat_history = []

def save_chat():
    with open(
        "chat_history.json",
        "w",
        encoding="utf-8"
    ) as file:

        json.dump(
            chat_history,
            file,
            indent=4,
            ensure_ascii=False
        )

# ==========================================
# SPEAK RESPONSE
# ==========================================
from IPython.display import Audio, display
def speak(text):

    tts = gTTS(
        text=text,
        lang="en"
    )

    tts.save("response.mp3")

    display(
        Audio(
            "response.mp3",
            autoplay=True
        )
    )

# ==========================================
# CAREER GUIDANCE
# ==========================================

def career_guidance():

    print("\nCareer Guidance Mode")
    print("Type exit to return")

    while True:

        user_input = input("\nYou: ")

        if user_input.lower() == "exit":
            break

        prompt = f"""
        You are Enisteen AI Career Mentor.

        Help students with:
        - Data Science
        - Machine Learning
        - Deep Learning
        - NLP
        - LLM
        - Career Planning
        - Skill Development

        Question:
        {user_input}
        """

        response = llm.invoke(prompt)

        answer = response.content

        print("\nEnisteen:")
        print(answer)

        chat_history.append({
            "mode":"career_guidance",
            "user":user_input,
            "assistant":answer
        })

# ==========================================
# ROADMAP GENERATOR
# ==========================================

def roadmap_generator():

    role = input(
        "\nEnter Career Goal: "
    )

    prompt = f"""
    Create a detailed roadmap
    for becoming a {role}.

    Include:

    1. Skills
    2. Tools
    3. Projects
    4. Learning Sequence
    5. Interview Preparation
    """

    response = llm.invoke(prompt)

    answer = response.content

    print("\nRoadmap:\n")
    print(answer)

    chat_history.append({
        "mode":"roadmap",
        "role":role,
        "roadmap":answer
    })

# ==========================================
# MOCK INTERVIEW
# ==========================================

def mock_interview():

    domain = input(
        "\nInterview Domain: "
    )

    print(
        f"\nStarting {domain} Interview"
    )

    for i in range(1,6):

        question_prompt = f"""
        Ask one interview question
        for {domain}.

        Question number {i}.
        """

        question = llm.invoke(
            question_prompt
        ).content

        print(
            f"\nQuestion {i}:"
        )

        print(question)

        answer = input(
            "\nYour Answer: "
        )

        evaluation_prompt = f"""
        Evaluate this answer.

        Question:
        {question}

        Candidate Answer:
        {answer}

        Give:

        1. Score out of 10
        2. Strengths
        3. Improvements
        """

        feedback = llm.invoke(
            evaluation_prompt
        ).content

        print("\nEvaluation:")
        print(feedback)

        chat_history.append({

            "mode":"mock_interview",

            "question":question,

            "candidate_answer":answer,

            "feedback":feedback
        })

# ==========================================
# VOICE CAREER ASSISTANT
# ==========================================

def voice_assistant():

    recognizer = sr.Recognizer()

    print(
        "\nVoice Assistant Started"
    )

    print(
        "Say exit to stop"
    )

    while True:

        try:

            with sr.Microphone() as source:

                print(
                    "\nListening..."
                )

                recognizer.adjust_for_ambient_noise(
                    source
                )

                audio = recognizer.listen(
                    source
                )

                user_text = recognizer.recognize_google(
                    audio
                )

                print(
                    "\nYou:",
                    user_text
                )

                if (
                    user_text.lower()
                    ==
                    "exit"
                ):
                    break

                prompt = f"""
                You are an AI Career Mentor.

                User Question:
                {user_text}
                """

                response = llm.invoke(
                    prompt
                )

                answer = response.content

                print(
                    "\nEnisteen:"
                )

                print(answer)

                speak(answer)

                chat_history.append({

                    "mode":"voice_assistant",

                    "user":user_text,

                    "assistant":answer
                })

        except Exception as e:

            print(
                "Error:",
                e
            )


# ==========================================
# VOICE MOCK INTERVIEW
# ==========================================

# ==========================================
# VOICE MOCK INTERVIEW
# ==========================================

def voice_mock_interview():

    recognizer = sr.Recognizer()

    domain = input(
        "\nInterview Domain: "
    )

    print(
        f"\nStarting Voice Interview for {domain}"
    )

    for i in range(1,6):

        question_prompt = f"""
        Ask one interview question for {domain}.
        Question number {i}.
        """

        question = llm.invoke(question_prompt).content

        print(f"\nQuestion {i}:")
        print(question)

        speak(question)

        try:

            with sr.Microphone() as source:

                print("\nSpeak your answer...")

                recognizer.adjust_for_ambient_noise(source)

                audio = recognizer.listen(source)

                candidate_answer = recognizer.recognize_google(audio)

                print("\nYour Answer:")
                print(candidate_answer)

        except Exception as e:

            print("Speech Error:", e)
            continue

        evaluation_prompt = f"""
        Evaluate this answer.

        Question:
        {question}

        Candidate Answer:
        {candidate_answer}

        Give:
        1. Score out of 10
        2. Strengths
        3. Improvements
        """

        feedback = llm.invoke(evaluation_prompt).content

        print("\nEvaluation:")
        print(feedback)

        speak(feedback)

        chat_history.append({

            "mode":"voice_mock_interview",

            "question":question,

            "candidate_answer":candidate_answer,

            "feedback":feedback
        })


# ==========================================
# MAIN MENU
# ==========================================

while True:

    print("\n")
    print("="*50)
    print("ENISTEEN AI INTERVIEW & CAREER COACH")
    print("="*50)

    print("1. Career Guidance")
    print("2. Roadmap Generator")
    print("3. Mock Interview")
    print("4. Voice Assistant")
    print("5. Voice Mock Interview")
    print("6. Save & Exit")
    choice = input(
        "\nSelect Option: "
    )

    if choice == "1":

        career_guidance()

    elif choice == "2":

        roadmap_generator()

    elif choice == "3":

        mock_interview()

    elif choice == "4":

        voice_assistant()

    elif choice == "5":

        voice_mock_interview()

    elif choice == "6":

        save_chat()

        print(
            "\nChat History Saved"
        )

        break

    else:

        print(
            "\nInvalid Choice"
        )
```
